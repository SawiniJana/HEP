```python
import uproot
import pandas as pd
import matplotlib.pyplot as plt
import awkward as ak
```

```python
f = "/home/sawini-jana/hep/MG5_aMC_v3_7_1/dihiggs_output.root"
file = uproot.open(f)
print("Available keys:", file.keys())
tree = file["Delphes"]
print("Branches:", tree.keys())
```

```python
branches = file['Delphes'].keys()
print(len(branches))
```

```python
pt = tree['Particle/Particle.PT']

plt.figure()
plt.hist(ak.flatten(pt.array()),bins=100, range = (500, 2000));
plt.xlabel(r'Particle $p_T$ (GeV)',fontsize=14)
plt.show()
```

```python
jet_pt   = tree["Jet.PT"]
jet_eta  = tree["Jet.Eta"]
jet_phi  = tree["Jet.Phi"]
jet_mass = tree["Jet.Mass"]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].hist(ak.flatten(jet_pt.array()),bins=100)
axes[0,0].set_xlabel(r'Jet $p_T$ (GeV)',fontsize=14)
axes[0,0].set_ylabel("Entries")

axes[0,1].hist(ak.flatten(jet_eta.array()),bins=100)
axes[0,1].set_xlabel(r'Jet ${\eta}$',fontsize=14)
axes[0,1].set_ylabel("Entries")

axes[1,0].hist(ak.flatten(jet_phi.array()),bins=100)
axes[1,0].set_xlabel(r'Jet ${\phi}$',fontsize=14)
axes[1,0].set_ylabel("Entries")

axes[1,1].hist(ak.flatten(jet_mass.array()),bins=100)
axes[1,1].set_xlabel(r'Jet $M$ (GeV)',fontsize=14)
axes[1,1].set_ylabel("Entries")

plt.tight_layout()
plt.show()
```

```python
fat_pt   = tree["FatJet.PT"]
fat_eta  = tree["FatJet.Eta"]
fat_phi  = tree["FatJet.Phi"]
fat_mass = tree["FatJet.Mass"]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].hist(ak.flatten(fat_pt.array()),bins=100)
axes[0,0].set_xlabel(r'Fat Jet $p_T$ (GeV)',fontsize=14)
axes[0,0].set_ylabel("Entries")

axes[0,1].hist(ak.flatten(fat_eta.array()),bins=100)
axes[0,1].set_xlabel(r'Fat Jet ${\eta}$',fontsize=14)
axes[0,1].set_ylabel("Entries")

axes[1,0].hist(ak.flatten(fat_phi.array()),bins=100)
axes[1,0].set_xlabel(r'Fat Jet ${\phi}$',fontsize=14)
axes[1,0].set_ylabel("Entries")

axes[1,1].hist(ak.flatten(fat_mass.array()),bins=100)
axes[1,1].set_xlabel(r'Fat Jet $M$ (GeV)',fontsize=14)
axes[1,1].set_ylabel("Entries")

plt.tight_layout()
plt.show()
```

```python
part_pt  = tree["Particle.PT"]
part_eta = tree["Particle.Eta"]
part_phi = tree["Particle.Phi"]
part_mass = tree["Particle.Mass"]


fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].hist(ak.flatten(part_pt.array()),bins=100)
axes[0,0].set_xlabel(r'Particle $p_T$ (GeV)',fontsize=14)
axes[0,0].set_ylabel("Entries")

axes[0,1].hist(ak.flatten(part_eta.array()),bins=100)
axes[0,1].set_xlabel(r'Particle ${\eta}$',fontsize=14)
axes[0,1].set_ylabel("Entries")

axes[1,0].hist(ak.flatten(part_phi.array()),bins=100)
axes[1,0].set_xlabel(r'Particle ${\phi}$',fontsize=14)
axes[1,0].set_ylabel("Entries")

axes[1,1].hist(ak.flatten(part_mass.array()),bins=100)
axes[1,1].set_xlabel(r'Particle $M$ (GeV)',fontsize=14)
axes[1,1].set_ylabel("Entries")

plt.tight_layout()
plt.show()
```

```python
pid = ak.flatten(tree["Particle.PID"].array())
bmask = abs(pid) == 5

b_pt  = ak.flatten(part_pt.array())[bmask]
b_eta = ak.flatten(part_eta.array())[bmask]
b_phi = ak.flatten(part_phi.array())[bmask]
b_mass = ak.flatten(part_mass.array())[bmask]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].hist(b_pt,bins=100)
axes[0,0].set_xlabel(r'B $p_T$ (GeV)',fontsize=14)
axes[0,0].set_ylabel("Entries")

axes[0,1].hist(b_eta,bins=100)
axes[0,1].set_xlabel(r'B ${\eta}$',fontsize=14)
axes[0,1].set_ylabel("Entries")

axes[1,0].hist(b_phi,bins=100)
axes[1,0].set_xlabel(r'B ${\phi}$',fontsize=14)
axes[1,0].set_ylabel("Entries")

axes[1,1].hist(b_mass,bins=100)
axes[1,1].set_xlabel(r'B $M$ (GeV)',fontsize=14)
axes[1,1].set_ylabel("Entries")

plt.tight_layout()
plt.show()
```

```python
jet_pt   = ak.flatten(tree["Jet.PT"].array())
jet_btag = ak.flatten(tree["Jet.BTag"].array())

bjet_pt = jet_pt[jet_btag == 1]


plt.figure()
plt.hist(bjet_pt,bins=100, range=(500,2000))
plt.xlabel(r'BTag $p_T$ (GeV)',fontsize=14)
plt.show()
```

### Observable Analysis

```python
import pandas as pd

file = open("/home/sawini-jana/hep/pythia8316/coding/observables.csv", "r", newline='')
df = pd.read_csv(file)

print(df['phi1'][0:5] + 1)
print(df.loc[10:12]['phi2'])

higgs = df[df.label==1][['phi1', 'phi2', 'phi3', 'phi4']].values
qcd = df[df.label==0][['phi1', 'phi2', 'phi3', 'phi4']].values
print(higgs)
```

```python
print(len(higgs)+len(qcd))
print(len(qcd)/len(higgs)) #3:5 ratio
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt

#HIGGS
EECnarrow_h = higgs[:,0]
EECwide_h = higgs[:,1]
ecf2_h = higgs[:,2]
ecf3_h = higgs[:,3]

#QCD
EECnarrow_qcd = qcd[:,0]
EECwide_qcd = qcd[:,1]
ecf2_qcd = qcd[:,2]
ecf3_qcd = qcd[:,3]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].hist(ecf2_h, bins=80, density=True, alpha=0.6, label="Higgs", color="black")
axes[0,0].hist(ecf2_qcd, bins=80, density=True, alpha=0.6, label="QCD", color="orange")
axes[0,0].set_xlabel(r'$e_2^{(1)}$',fontsize=14)
axes[0,0].set_ylabel("Entries")
axes[0,0].legend()

axes[0,1].hist(ecf3_h, bins=80, density=True, alpha=0.6, label="Higgs", color="black")
axes[0,1].hist(ecf3_qcd, bins=80, density=True, alpha=0.6, label="QCD", color ="orange")
axes[0,1].set_xlabel(r'$e_3^{(1)}$',fontsize=14)
axes[0,1].set_ylabel("Entries")
axes[0,1].legend()

axes[1,0].hist(EECnarrow_h, bins=80, density=True, alpha=0.6, label="Higgs")
axes[1,0].hist(EECnarrow_qcd, bins=80, density=True, alpha=0.6, label="QCD")
axes[1,0].set_xlabel(r'EEC narrow',fontsize=14)
axes[1,0].set_ylabel("Entries")
axes[1,0].legend()

axes[1,1].hist(EECwide_h, bins=80, density=True, alpha=0.6, label="Higgs")
axes[1,1].hist(EECwide_qcd, bins=80, density=True, alpha=0.6, label="QCD")
axes[1,1].set_xlabel(r'EEC wide',fontsize=14)
axes[1,1].set_ylabel("Entries")
axes[1,1].legend()

plt.show()
```

```python
fig,ax=plt.subplots(2,2,figsize=(10,8))

ax[0,0].scatter(ecf2_h,ecf3_h,s=6,alpha=0.4,label="Higgs")
ax[0,0].scatter(ecf2_qcd,ecf3_qcd,s=6,alpha=0.4,label="QCD")
ax[0,0].set_xlabel(r"$e_2$")
ax[0,0].set_ylabel(r"$e_3$")
ax[0,0].legend()

ax[0,1].scatter(EECnarrow_h,EECwide_h,s=6,alpha=0.4,label="Higgs")
ax[0,1].scatter(EECnarrow_qcd,EECwide_qcd,s=6,alpha=0.4,label="QCD")
ax[0,1].set_xlabel("EEC narrow")
ax[0,1].set_ylabel("EEC wide")
ax[0,1].legend()

ax[1,0].scatter(ecf2_h,EECwide_h,s=6,alpha=0.4,label="Higgs")
ax[1,0].scatter(ecf2_qcd,EECwide_qcd,s=6,alpha=0.4,label="QCD")
ax[1,0].set_xlabel(r"$e_2$")
ax[1,0].set_ylabel("EEC wide")
ax[1,0].legend()

ax[1,1].scatter(ecf3_h,EECnarrow_h,s=6,alpha=0.4,label="Higgs")
ax[1,1].scatter(ecf3_qcd,EECnarrow_qcd,s=6,alpha=0.4,label="QCD")
ax[1,1].set_xlabel(r"$e_3$")
ax[1,1].set_ylabel("EEC narrow")
ax[1,1].legend()

plt.show()
```
